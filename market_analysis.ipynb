```python



def parse_saved_etsy_pages(folder="etsy_pages"):
    folder = Path(folder)
    files = sorted(
        p for p in folder.glob("*")
        if p.suffix.lower() in {".html", ".htm", ".txt"}
    )
    if not files:
        raise FileNotFoundError(f"No saved pages found in {folder.resolve()}")

    rows = []

    for path in files:
        soup = BeautifulSoup(path.read_bytes(), "html.parser")

        # Matches the actual card structure in your supplied HTML.
        cards = soup.select("[data-listing-card-v2][data-listing-id]")

        if not cards:
            warnings.warn(f"{path.name}: no listing cards found.")
            continue

        count = 0
        for card in cards:
            link = card.select_one('a.listing-link[href*="/listing/"]')
            if link is None:
                continue

            match = re.search(r"/listing/(\d+)", link["href"])
            if not match:
                continue

            heading = card.select_one("h3.v2-listing-card__title")
            image = card.select_one("img")
            price_area = card.select_one(".n-listing-card__price")
            price_text = (
                price_area.get_text(" ", strip=True) if price_area else ""
            )

            # Only include cards explicitly marked as sold.
            if not re.search(r"\bsold(?:\s+out)?\b", price_text, re.I):
                continue

            price = (
                price_area.select_one(".currency-value")
                if price_area else None
            )
            currency = (
                price_area.select_one(".currency-symbol")
                if price_area else None
            )

            rows.append({
                "listing_id": match.group(1),
                "shop_id": card.get("data-shop-id"),
                "title": (
                    link.get("title")
                    or (heading.get_text(" ", strip=True) if heading else None)
                    or (image.get("alt") if image else None)
                ),
                "status": price_text,
                "displayed_price": (
                    price.get_text(strip=True) if price else None
                ),
                "currency_symbol": (
                    currency.get_text(strip=True) if currency else None
                ),
                "listing_url": f"https://www.etsy.com/listing/{match.group(1)}",
                "image_url": (
                    image.get("data-src") or image.get("src")
                    if image else None
                ),
                "source_file": path.name,
            })
            count += 1

        # print(f"{path.name}: {count} sold cards")

    if not rows:
        raise RuntimeError("No cards explicitly marked Sold were found.")

    return pd.DataFrame(rows).convert_dtypes()





# Optional:
# summary_df.to_csv("sold_listing_summary.csv", index=False)
```

```python
from pathlib import Path
import textwrap

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
from IPython.display import display


def display_store_data(store, top_n=20):
    if top_n is not None and top_n < 1:
        raise ValueError("top_n must be positive or None.")

    df = parse_saved_etsy_pages(Path("etsy_pages") / store)

    summary_df = (
        df.groupby("listing_id", as_index=False)
        .agg(
            name=("title", "first"),
            times_observed=("listing_id", "size"),
        )
        .sort_values("times_observed", ascending=False)
        [["name", "times_observed"]]
        .reset_index(drop=True)
    )

    print(store)
    display(summary_df)

    plot_df = (
        summary_df.head(top_n) if top_n is not None else summary_df
    ).copy()

    if plot_df.empty:
        print("No listings to graph.")
        return summary_df

    labels = [
        textwrap.fill(str(name), width=48)
        for name in plot_df["name"]
    ]
    values = plot_df["times_observed"].to_numpy()
    positions = np.arange(len(plot_df))

    # Allow extra space for wrapped titles.
    height = max(
        5,
        2 + sum(label.count("\n") + 1 for label in labels) * 0.32,
    )

    with plt.rc_context({
        "font.family": "sans-serif",
        "font.size": 10,
        "axes.labelcolor": "#475569",
        "xtick.color": "#64748B",
        "ytick.color": "#334155",
    }):
        fig, ax = plt.subplots(
            figsize=(13, height),
            layout="constrained",
        )
        fig.set_facecolor("#F8FAFC")
        ax.set_facecolor("#F8FAFC")

        # Color intensity follows the observation count.
        scale = values / values.max()
        colors = plt.cm.Blues(0.35 + 0.45 * scale)
        colors[values == values.max()] = plt.matplotlib.colors.to_rgba(
            "#0D9488"
        )

        bars = ax.barh(
            positions,
            values,
            height=0.65,
            color=colors,
            edgecolor="none",
            zorder=3,
        )

        ax.set_yticks(positions, labels=labels)
        ax.invert_yaxis()
        ax.tick_params(axis="both", length=0, pad=9)

        ax.bar_label(
            bars,
            labels=[f"{int(value):,}" for value in values],
            padding=7,
            fontsize=10,
            fontweight="bold",
            color="#334155",
        )

        fig.suptitle(
            f"{store} | Sold listing appearances",
            fontsize=18,
            fontweight="bold",
            color="#0F172A",
            ha="left",
            x=0.02,
        )
        ax.set_title(
            f"Showing {len(plot_df):,} of {len(summary_df):,} unique listings"
            "  •  Teal marks the highest count",
            loc="left",
            fontsize=10,
            color="#64748B",
            pad=18,
        )

        ax.set_xlabel(
            "Appearances in saved pages · not verified sales",
            labelpad=14,
        )
        ax.set_xlim(0, values.max() * 1.16)
        ax.xaxis.set_major_locator(MaxNLocator(integer=True, nbins=6))

        ax.set_axisbelow(True)
        ax.grid(axis="x", color="#E2E8F0", linewidth=0.8)
        ax.grid(axis="y", visible=False)

        for spine in ax.spines.values():
            spine.set_visible(False)

        ax.margins(y=0.025)
        plt.show()

    # return summary_df


# Example:
# summary_df = display_store_data("MunioroPrints", top_n=20)
```
