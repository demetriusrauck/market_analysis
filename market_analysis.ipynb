```python
%pip -q install beautifulsoup4 pandas

from pathlib import Path
import re
import warnings

import pandas as pd
from bs4 import BeautifulSoup
from IPython.display import display
from matplotlib.ticker import MaxNLocator
import textwrap
%matplotlib inline
import matplotlib.pyplot as plt
```

```python



def parse_saved_etsy_pages(folder="etsy_pages"):
    folder = Path(folder)
    files = sorted(
        p for p in folder.glob("*")
        if p.suffix.lower() in {".html", ".htm", ".txt"}
    )
    if not files:
        raise FileNotFoundError(f"No saved pages found in {folder.resolve()}")

    rows = []

    for path in files:
        soup = BeautifulSoup(path.read_bytes(), "html.parser")

        # Matches the actual card structure in your supplied HTML.
        cards = soup.select("[data-listing-card-v2][data-listing-id]")

        if not cards:
            warnings.warn(f"{path.name}: no listing cards found.")
            continue

        count = 0
        for card in cards:
            link = card.select_one('a.listing-link[href*="/listing/"]')
            if link is None:
                continue

            match = re.search(r"/listing/(\d+)", link["href"])
            if not match:
                continue

            heading = card.select_one("h3.v2-listing-card__title")
            image = card.select_one("img")
            price_area = card.select_one(".n-listing-card__price")
            price_text = (
                price_area.get_text(" ", strip=True) if price_area else ""
            )

            # Only include cards explicitly marked as sold.
            if not re.search(r"\bsold(?:\s+out)?\b", price_text, re.I):
                continue

            price = (
                price_area.select_one(".currency-value")
                if price_area else None
            )
            currency = (
                price_area.select_one(".currency-symbol")
                if price_area else None
            )

            rows.append({
                "listing_id": match.group(1),
                "shop_id": card.get("data-shop-id"),
                "title": (
                    link.get("title")
                    or (heading.get_text(" ", strip=True) if heading else None)
                    or (image.get("alt") if image else None)
                ),
                "status": price_text,
                "displayed_price": (
                    price.get_text(strip=True) if price else None
                ),
                "currency_symbol": (
                    currency.get_text(strip=True) if currency else None
                ),
                "listing_url": f"https://www.etsy.com/listing/{match.group(1)}",
                "image_url": (
                    image.get("data-src") or image.get("src")
                    if image else None
                ),
                "source_file": path.name,
            })
            count += 1

        # print(f"{path.name}: {count} sold cards")

    if not rows:
        raise RuntimeError("No cards explicitly marked Sold were found.")

    return pd.DataFrame(rows).convert_dtypes()





# Optional:
# summary_df.to_csv("sold_listing_summary.csv", index=False)
```

```python
def display_store_data(store):
    print(store)
    df = parse_saved_etsy_pages("etsy_pages/"+store)
    summary_df = (
        df.groupby("listing_id", as_index=False)
        .agg(
            name=("title", "first"),
            times_observed=("listing_id", "size"),
        )
        .sort_values("times_observed", ascending=False)
        [["name", "times_observed"]]
        .reset_index(drop=True)
    )

    # with pd.option_context(
    #     "display.max_rows", None,
    #     "display.max_columns", None,
    #     "display.max_colwidth", None,
    # ):
    display(summary_df)
    # summary_df.head(50)
    # return summary_df
```

```python
print("MunioroPrints")
df_MunioroPrints = parse_saved_etsy_pages("etsy_pages/MunioroPrints")
summary_df_MunioroPrints = (
    df_MunioroPrints.groupby("listing_id", as_index=False)
      .agg(
          name=("title", "first"),
          times_observed=("listing_id", "size"),
      )
      .sort_values("times_observed", ascending=False)
      [["name", "times_observed"]]
      .reset_index(drop=True)
)

# with pd.option_context(
#     "display.max_rows", None,
#     "display.max_columns", None,
#     "display.max_colwidth", None,
# ):
#     display(summary_df_MunioroPrints)


plot_df = summary_df_MunioroPrints.sort_values(
    "times_observed", ascending=False
).reset_index(drop=True)

if plot_df.empty:
    print("No listings to graph.")
else:
    labels = [
        textwrap.fill(str(name), width=55)
        for name in plot_df["name"]
    ]

    fig, ax = plt.subplots(
        figsize=(12, max(5, sum(label.count("\n") + 1 for label in labels) * 0.3)),
        layout="constrained",
    )

    bars = ax.barh(
        range(len(plot_df)),
        plot_df["times_observed"],
        color="steelblue",
    )

    ax.set_yticks(range(len(plot_df)), labels=labels)
    ax.invert_yaxis()
    ax.bar_label(bars, padding=4, fmt="%.0f")

    ax.set_title("MunioroPrints — Sold Listing Appearances")
    ax.set_xlabel("Times observed in saved pages")
    ax.xaxis.set_major_locator(MaxNLocator(integer=True))
    ax.set_xlim(0, plot_df["times_observed"].max() * 1.15)
    ax.set_axisbelow(True)
    ax.grid(axis="x", alpha=0.25)
    ax.spines[["top", "right"]].set_visible(False)

    plt.show()

```

```python
%matplotlib inline
import matplotlib.pyplot as plt
from IPython.display import display

top = (
    summary_df_MunioroPrints
    .nlargest(10, "times_observed")
    .sort_values("times_observed")
)

fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(range(len(top)), top["times_observed"], color="steelblue")
ax.set_yticks(
    range(len(top)),
    [str(name)[:45] for name in top["name"]],
)
ax.set_title("MunioroPrints — Top 10 Listing Appearances")
ax.set_xlabel("Times observed")
fig.tight_layout()

display(fig)
plt.close(fig)
```
